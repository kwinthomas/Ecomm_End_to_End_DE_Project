```python
%sql
CREATE TABLE IF NOT EXISTS ecomm.bronze.orders_raw
USING PARQUET
LOCATION 'abfss://basecontainer@ecommprojadls.dfs.core.windows.net/bronze/orders/';
```

```python
%sql
CREATE TABLE IF NOT EXISTS ecomm.silver.orders_current (
    order_id                      STRING NOT NULL,
    customer_unique_id            STRING,
    order_status                  STRING,
    order_purchase_timestamp      TIMESTAMP,
    order_approved_at             TIMESTAMP,
    order_delivered_carrier_date  TIMESTAMP,
    order_delivered_customer_date TIMESTAMP,
    order_estimated_delivery_date TIMESTAMP,
    updated_at                    TIMESTAMP,
    _change_version               BIGINT,
    _is_deleted                   BOOLEAN,
    _merged_at                    TIMESTAMP
) USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS ecomm.silver.customers_current (
    customer_unique_id  STRING NOT NULL,
    customer_zip_prefix STRING,
    customer_city       STRING,
    customer_state      STRING,
    updated_at          TIMESTAMP,
    _change_version     BIGINT,
    _is_deleted         BOOLEAN,
    _merged_at          TIMESTAMP
) USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS ecomm.gold.dim_customer (
    customer_sk         STRING,
    customer_unique_id  STRING,
    customer_city       STRING,
    customer_state      STRING,
    customer_zip_prefix STRING,
    valid_from          TIMESTAMP,
    valid_to            TIMESTAMP,
    is_current          BOOLEAN
) USING DELTA;
```
