```python
%sql
CREATE OR REPLACE TABLE ecomm.gold.fact_order_lifecycle AS
SELECT
    order_id,
    customer_unique_id,
    order_status,
    order_purchase_timestamp,
    order_approved_at,
    order_delivered_carrier_date,
    order_delivered_customer_date,
    order_estimated_delivery_date,
    TIMESTAMPDIFF(HOUR, order_purchase_timestamp, order_approved_at) AS hrs_to_approve,
    TIMESTAMPDIFF(HOUR, order_approved_at, order_delivered_carrier_date) AS hrs_to_ship,
    TIMESTAMPDIFF(DAY, order_delivered_carrier_date, order_delivered_customer_date) AS days_in_transit,
    TIMESTAMPDIFF(DAY, order_purchase_timestamp, order_delivered_customer_date) AS days_total,
    TIMESTAMPDIFF(DAY, order_estimated_delivery_date, order_delivered_customer_date) AS days_vs_estimate,
    order_delivered_customer_date <= order_estimated_delivery_date AS delivered_on_time,
    _is_deleted AS is_cancelled
FROM ecomm.silver.orders_current;
```
